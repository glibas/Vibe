## 📁 Dataset Setup

```python
from google.colab import drive
import shutil
import zipfile
import os

# Mount Google Drive
drive.mount('/content/drive')

# Define paths for WebDesignPrototypicality dataset
zip_path = '/content/drive/MyDrive/datasets/webdesignprototypicality.zip'
local_zip = '/content/webdesignprototypicality.zip'
extract_dir = '/content/webdesignprototypicality'

if not os.path.exists(extract_dir):
    # Copy zip from Drive to Colab local storage
    shutil.copy(zip_path, local_zip)
    # Extract dataset locally
    os.makedirs(extract_dir, exist_ok=True)
    with zipfile.ZipFile(local_zip, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print("📊 Dataset extracted to:", extract_dir)
else:
    print("📊 Dataset already available at:", extract_dir)

# List dataset contents
!ls -la {extract_dir}
```

## 🧠 Model Creation and Basic Usage

```python
import torch
from vibe.models import BeautyPredictor
from vibe.data import InterfaceDataLoader

# Create model with saliency guidance
model = BeautyPredictor(
    model_type='saliency_guided',
    img_size=224,
    embed_dim=768,
    n_layers=12,
    n_heads=12,
    use_beauty_tokens=True,
    saliency_weight=0.5
)

print(f"🎯 Model created with {sum(p.numel() for p in model.parameters()):,} parameters")
print(f"🔧 Model type: {model.model_type}")
print(f"🎨 Beauty tokens enabled: {model.use_beauty_tokens}")
```

## 🎨 Interface Beauty Analysis Demo

```python
# Create synthetic interface for demonstration
synthetic_interface = torch.randn(1, 3, 224, 224)

# Predict beauty score
model.eval()
with torch.no_grad():
    beauty_score = model.predict_beauty_score(synthetic_interface)
    print(f"🎯 Beauty Score: {beauty_score:.3f}")
    
    # Comprehensive analysis
    analysis = model.analyze_interface(synthetic_interface)
    explanation = model.get_beauty_explanation(analysis)
    
    print("\n📊 Comprehensive Analysis:")
    print("="*50)
    print(explanation)
```
